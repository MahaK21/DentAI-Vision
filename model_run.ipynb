```python
import os

def read_first_chars(file_path):
    """Reads the first character of each line in a file and stores it in a list."""
    first_chars = []
    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            for line in file:
                if line:  # Ensure the line is not empty
                    first_chars.append(line[0])
    except Exception as e:
        print(f"Error reading {file_path}: {e}")
    return first_chars

def process_folder(folder_path):
    """Iterates through all files in a folder and aggregates first character lists."""
    aggregated_lists = []
    for filename in os.listdir(folder_path):
        file_path = os.path.join(folder_path, filename)
        if os.path.isfile(file_path):
            first_chars = read_first_chars(file_path)
            aggregated_lists.append(first_chars)
    return aggregated_lists

def filter_lists_by_char(aggregated_lists, search_char):
    """Finds all lists that contain at least one occurrence of the search character."""
    return [char_list for char_list in aggregated_lists if search_char in char_list]
```

```python
import os
import cv2

# Define paths
train_img_dir = "/home/kam/Documents/ELEC278/Qmind/Dataset/YoloData3/images/train"
train_label_dir = "/home/kam/Documents/ELEC278/Qmind/Dataset/YoloData3/labels/train"
val_img_dir = "/home/kam/Documents/ELEC278/Qmind/Dataset/YoloData3/images/val"
val_label_dir = "/home/kam/Documents/ELEC278/Qmind/Dataset/YoloData3/labels/val"

# Function to mirror images and adjust labels
def mirror_dataset(img_dir, label_dir):
    for filename in os.listdir(img_dir):
        if filename.endswith(".jpg") or filename.endswith(".png"):
            # Load image
            img_path = os.path.join(img_dir, filename)
            img = cv2.imread(img_path)

            # Flip image over Y-axis
            mirrored_img = cv2.flip(img, 1)

            # Save mirrored image
            mirrored_img_path = os.path.join(img_dir, "mirrored_" + filename)
            cv2.imwrite(mirrored_img_path, mirrored_img)

            # Process YOLO label
            label_filename = filename.replace(".jpg", ".txt").replace(".png", ".txt")
            label_path = os.path.join(label_dir, label_filename)

            if os.path.exists(label_path):
                with open(label_path, "r") as f:
                    lines = f.readlines()

                new_lines = []
                for line in lines:
                    values = line.strip().split()
                    class_id = values[0]
                    coords = list(map(float, values[1:]))

                    # If it's a bounding box format
                    if len(coords) == 4:
                        x_center, y_center, width, height = coords
                        x_center_new = 1 - x_center  # Mirror over Y-axis
                        new_lines.append(f"{class_id} {x_center_new:.6f} {y_center:.6f} {width:.6f} {height:.6f}")

                    # If it's a segmentation format
                    elif len(coords) % 2 == 0:
                        new_coords = []
                        for i in range(0, len(coords), 2):
                            x, y = coords[i], coords[i + 1]
                            x_new = 1 - x  # Flip x-coordinate
                            new_coords.append(f"{x_new:.6f} {y:.6f}")
                        new_lines.append(f"{class_id} " + " ".join(new_coords))

                # Save mirrored label
                mirrored_label_path = os.path.join(label_dir, "mirrored_" + label_filename)
                with open(mirrored_label_path, "w") as f:
                    f.write("\n".join(new_lines) + "\n")

# Apply mirroring to training and validation datasets
mirror_dataset(train_img_dir, train_label_dir)
mirror_dataset(val_img_dir, val_label_dir)

print("Mirroring complete! New images and labels saved.")
```

```python
import os
import cv2

# Define paths

# Function to ensure directory exists
def ensure_dir(directory):
    if not os.path.exists(directory):
        os.makedirs(directory)

# Function to mirror images and adjust labels
def mirror_dataset(img_dir, label_dir, output_img_dir, output_label_dir):
    ensure_dir(output_img_dir)
    ensure_dir(output_label_dir)

    for filename in os.listdir(img_dir):
        if filename.endswith(".jpg") or filename.endswith(".png"):
            # Load image
            img_path = os.path.join(img_dir, filename)
            img = cv2.imread(img_path)

            if img is None:
                print(f"Error loading image: {img_path}")
                continue

            # Flip image over Y-axis
            mirrored_img = cv2.flip(img, 1)

            # Save mirrored image
            mirrored_filename = "mirrored_" + filename
            mirrored_img_path = os.path.join(output_img_dir, mirrored_filename)
            cv2.imwrite(mirrored_img_path, mirrored_img)

            # Process YOLO label
            label_filename = filename.replace(".jpg", ".txt").replace(".png", ".txt")
            label_path = os.path.join(label_dir, label_filename)

            if os.path.exists(label_path):
                with open(label_path, "r") as f:
                    lines = f.readlines()

                new_lines = []
                for line in lines:
                    values = line.strip().split()
                    class_id = values[0]
                    coords = list(map(float, values[1:]))

                    # If it's a bounding box format
                    if len(coords) == 4:
                        x_center, y_center, width, height = coords
                        x_center_new = 1 - x_center  # Mirror over Y-axis
                        new_lines.append(f"{class_id} {x_center_new:.6f} {y_center:.6f} {width:.6f} {height:.6f}")

                    # If it's a segmentation format
                    elif len(coords) % 2 == 0:
                        new_coords = []
                        for i in range(0, len(coords), 2):
                            x, y = coords[i], coords[i + 1]
                            x_new = 1 - x  # Flip x-coordinate
                            new_coords.append(f"{x_new:.6f} {y:.6f}")
                        new_lines.append(f"{class_id} " + " ".join(new_coords))

                # Save mirrored label
                mirrored_label_path = os.path.join(output_label_dir, mirrored_filename.replace(".jpg", ".txt"))
                with open(mirrored_label_path, "w") as f:
                    f.write("\n".join(new_lines) + "\n")

            print(f"✅ Mirrored image and label saved: {mirrored_img_path}")

# Create mirrored dataset
mirror_dataset(train_img_dir, train_label_dir, "train/mirrored_images", "train/mirrored_labels")
mirror_dataset(val_img_dir, val_label_dir, "val/mirrored_images", "val/mirrored_labels")

print("🎉 Mirroring complete! New images and labels saved.")
```

```python
import os
import cv2
import numpy as np
from glob import glob
import shutil

# Paths to original dataset
train_images_path = "/home/kam/Documents/ELEC278/Qmind/Dataset/YoloData3/images/train"
train_labels_path = "/home/kam/Documents/ELEC278/Qmind/Dataset/YoloData3/labels/train"

# New dataset directory (combined original + noisy data)
output_images_path = "/home/kam/Documents/ELEC278/Qmind/Dataset/YoloData4/images"
output_labels_path = "/home/kam/Documents/ELEC278/Qmind/Dataset/YoloData4/labels"

# Ensure output directories exist
os.makedirs(output_images_path, exist_ok=True)
os.makedirs(output_labels_path, exist_ok=True)

def add_gaussian_noise(image, mean=0, std=25):
    """Add Gaussian noise to an image."""
    noise = np.random.normal(mean, std, image.shape).astype(np.uint8)
    noisy_image = cv2.add(image, noise)
    return noisy_image

def add_salt_and_pepper_noise(image, salt_prob=0.02, pepper_prob=0.02):
    """Add Salt and Pepper noise to an image."""
    noisy_image = image.copy()
    total_pixels = image.size
    num_salt = int(salt_prob * total_pixels)
    num_pepper = int(pepper_prob * total_pixels)

    # Add Salt noise (white pixels)
    salt_coords = [np.random.randint(0, i - 1, num_salt) for i in image.shape[:2]]
    noisy_image[salt_coords[0], salt_coords[1]] = 255

    # Add Pepper noise (black pixels)
    pepper_coords = [np.random.randint(0, i - 1, num_pepper) for i in image.shape[:2]]
    noisy_image[pepper_coords[0], pepper_coords[1]] = 0

    return noisy_image

# Process images
for image_path in glob(os.path.join(train_images_path, "*.jpg")):
    filename = os.path.basename(image_path)
    label_path = os.path.join(train_labels_path, filename.replace(".jpg", ".txt"))

    # Load the image
    image = cv2.imread(image_path)

    # Save the original image to new dataset
    shutil.copy(image_path, os.path.join(output_images_path, filename))

    # Apply noise augmentation
    noisy_image = add_gaussian_noise(image)
    noisy_image = add_salt_and_pepper_noise(noisy_image)

    # Save the noisy image with a new name
    noisy_filename = filename.replace(".jpg", "_noisy.jpg")
    cv2.imwrite(os.path.join(output_images_path, noisy_filename), noisy_image)

    # Copy the original label file
    if os.path.exists(label_path):
        shutil.copy(label_path, os.path.join(output_labels_path, filename.replace(".jpg", ".txt")))
        shutil.copy(label_path, os.path.join(output_labels_path, noisy_filename.replace(".jpg", ".txt")))

print("Dataset augmentation completed successfully!")
```
